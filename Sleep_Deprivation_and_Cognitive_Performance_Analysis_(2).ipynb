```python
import matplotlib.pyplot as plt

plt.hist(df["Sleep_Hours"])

plt.title("Distribution of Sleep Hours")
plt.xlabel("Sleep Hours")
plt.ylabel("Number of Participants")

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.scatter(df["Sleep_Hours"], df["N_Back_Accuracy"])

plt.title("Sleep Hours vs N-Back Accuracy")
plt.xlabel("Sleep Hours")
plt.ylabel("N-Back Accuracy")

plt.show()
```

```python
!pip install seaborn
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

correlation_matrix = df.corr(numeric_only=True)

plt.figure(figsize=(10, 8))

sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```

```python
sns.regplot(x="Sleep_Hours", y="N_Back_Accuracy", data=df)

plt.title("Sleep Hours vs N-Back Accuracy")
plt.xlabel("Sleep Hours")
plt.ylabel("N-Back Accuracy")

plt.show()
```

```python
sns.boxplot(y=df["Sleep_Hours"])

plt.title("Boxplot of Sleep Hours")

plt.show()
```

```python
sns.boxplot(x="Gender", y="Sleep_Hours", data=df)

plt.title("Sleep Hours by Gender")

plt.show()
```

```python
sns.pairplot(df[
    [
        "Sleep_Hours",
        "Sleep_Quality_Score",
        "N_Back_Accuracy",
        "Stress_Level"
    ]
])

plt.show()
```

```python
df.groupby("Gender")[["Sleep_Hours", "N_Back_Accuracy", "Stress_Level"]].mean()
```

```python
sns.barplot(x="Gender", y="Sleep_Hours", data=df)

plt.title("Average Sleep Hours by Gender")

plt.show()
```

```python
sns.barplot(x="Gender", y="Stress_Level", data=df)

plt.title("Average Stress Level by Gender")

plt.show()
```

# Conclusions

The analysis explored relationships between sleep duration, cognitive performance, and stress levels.

The correlation analysis showed a weak negative relationship between sleep hours and N-Back accuracy (correlation coefficient = -0.11), suggesting that sleep duration had little effect on cognitive performance in this dataset.

Descriptive statistics indicated that the average sleep duration among participants was approximately 5.8 hours, while the average stress level was moderate.

The t-test comparing sleep duration between males and females revealed no statistically significant differences (p > 0.05).

Visualizations such as histograms, scatterplots, heatmaps, boxplots, and pairplots helped identify general trends and distributions within the data.

Overall, the dataset did not show strong relationships between the analyzed variables, but the analysis demonstrated important data analysis and visualization techniques using Python.
